```python
from sklearn.preprocessing import OneHotEncoder

categorical_cols = [
    "product_type_name",
    "product_group_name",
    "graphical_appearance_name",
    "colour_group_name",
    "perceived_colour_value_name",
    "perceived_colour_master_name",
    "department_name",
    "index_name",
    "index_group_name",
    "section_name",
    "garment_group_name"
]

encoder = OneHotEncoder(
    handle_unknown="ignore",
    sparse_output=True
)

categorical_matrix = encoder.fit_transform(articles[categorical_cols])

categorical_matrix.shape
```

```python
color_cols = [
    "colour_group_name",
    "perceived_colour_value_name",
    "perceived_colour_master_name"
]

articles[color_cols].nunique()
```

```python
color_mapping_check = articles.groupby("colour_group_name", observed=True).agg(
    unique_value_count=("perceived_colour_value_name", "nunique"),
    unique_master_count=("perceived_colour_master_name", "nunique")
)

color_mapping_check.sort_values(
    ["unique_value_count", "unique_master_count"],
    ascending=False
).head(20)
```
