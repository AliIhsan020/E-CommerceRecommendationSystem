## EDA Notes — Column Types

These columns are int64 but categorical (code/ID nature, no mathematical meaning):
product_type_no, graphical_appearance_no, colour_group_code,
perceived_colour_value_id, perceived_colour_master_id,
department_no, index_group_no, section_no, garment_group_no

→ Convert to category type in cleaning step

```python
cat_cols = [
    'product_type_no', 'graphical_appearance_no', 'colour_group_code',
    'perceived_colour_value_id', 'perceived_colour_master_id',
    'department_no', 'index_group_no', 'section_no', 'garment_group_no'
]

for col in cat_cols:
    top = df[col].value_counts(normalize=True).iloc[0] * 100
    print(f"{col}: most dominant category %{top:.1f}")
```
