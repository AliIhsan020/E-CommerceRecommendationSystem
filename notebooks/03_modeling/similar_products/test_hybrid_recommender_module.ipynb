```python
from pathlib import Path
import sys

# Notebook farklı çalışma dizinlerinden açılırsa import path bozulmasın diye
PROJECT_ROOT = Path("../../../").resolve()

if not (PROJECT_ROOT / "src").exists():
    PROJECT_ROOT = Path(".").resolve()

sys.path.append(str(PROJECT_ROOT))

import pandas as pd

from src.recommenders.similar_products.hybrid_recommender import (
    HybridSimilarProductsRecommender
)
```

```python
articles = pd.read_parquet(
    PROJECT_ROOT / "data/gold/articles.parquet"
)

articles.shape
```

```python
recommender = HybridSimilarProductsRecommender(
    categorical_weight=1.5,
    text_weight=1.0,
    max_features=50000,
    candidate_pool=200
)

recommender.fit(articles)
```

```python
query_product, recommendations = recommender.recommend(
    article_id=108775015,
    top_k=10
)

query_product, recommendations
```

```python
model_path = PROJECT_ROOT / "artifacts/models/similarproducts/hybrid"

recommender.save(model_path)

print("Model saved to:", model_path)
```

```python
loaded_recommender = HybridSimilarProductsRecommender.load(model_path)

query_product_loaded, recommendations_loaded = loaded_recommender.recommend(
    article_id=108775015,
    top_k=10
)

query_product_loaded, recommendations_loaded
```

```python
recommendations[["article_id", "rank"]].reset_index(drop=True).equals(
    recommendations_loaded[["article_id", "rank"]].reset_index(drop=True)
)
```
